```python
from collections import deque
import time
#from plotly import graph_objects as go
#import numpy as np
#from typing import List
from functools import cache

day = 12

test_result = 2
test_data = """\
0:
###
##.
##.

1:
###
##.
.##

2:
.##
###
##.

3:
##.
###
##.

4:
###
#..
###

5:
###
.#.
###

4x4: 0 0 0 0 2 0
12x5: 1 0 1 0 2 2
12x5: 1 0 1 0 3 2
"""
```

```python
# solution does not need packing gifts together just checking if enough space is available...

def solution1(data):
    *gifts, regions = data.split("\n\n")
    shapes = {}
    for g in gifts:
        key, *shape = g.split("\n")
        key = int(key.strip(":"))
        shape = [[int(sym=="#") for sym in line] for line in shape]
        shapes[key] = shape
    
    dims = []
    counts = []
    for line in regions.split("\n")[:-1]:
        d, c = line.split(": ")
        dims.append(list(map(int, d.split("x"))))
        counts.append(list(map(int, c.split(" "))))

    output = 0 
    for count, dim in zip(counts, dims):
        space_needed = 0
        for i, c in enumerate(count):
            if c > 0:
                space_needed += sum(map(sum, shapes[i]))*c
        x, y = dim
        if space_needed <= x*y:
            output += 1
    return output

    
sol1 = solution1(test_data)
print(sol1)
#assert sol1 == test_result
# working solution does not work on example
print("test passed")
```

```python
with open(f"Day{day:02d}.txt") as f:
    inp_data = f.read()
t0 = time.time()
sol1 = solution1(inp_data)
print(f"elapsed time {time.time()-t0:.2f} s")
print(sol1)
assert sol1 == 579
print("test passed")
```
