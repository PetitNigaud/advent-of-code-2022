### Part 1

```python
from collections import deque
import time
#from plotly import graph_objects as go
#import numpy as np
#from typing import List
from functools import cache

day = 11

test_result = 5
test_data = """\
aaa: you hhh
you: bbb ccc
bbb: ddd eee
ccc: ddd eee fff
ddd: ggg
eee: out
fff: out
ggg: out
hhh: ccc fff iii
iii: out
"""
```

```python
def recurse(start: str, data: dict):
    #print(f"recurse({start})")
    global output
    for out in data.get(start, []):
        #print(f"{start} -> {out}")
        if out == "out":
            #print(output)
            output += 1
        else:
            recurse(out, data)

output = 0
def solution1(data):
    global output
    lines = data.split("\n")[:-1]    
    output = 0
    data = {}
    for line in lines:
        key, *values = line.split(" ")
        data[key[:-1]] = values
    
    output = 0
   
    recurse("you", data)
    return output

    
sol1 = solution1(test_data)
print(sol1)
assert sol1 == test_result
print("test passed")
```

```python
with open(f"Day{day:02d}.txt") as f:
    inp_data = f.read()
t0 = time.time()
sol1 = solution1(inp_data)
print(f"elapsed time {time.time()-t0:.2f} s")
print(sol1)
assert sol1 == 719
print("test passed")
```

```python
test_result2 = 2

test_data = """\
svr: aaa bbb
aaa: fft
fft: ccc
bbb: tty
tty: ccc
ccc: ddd eee
ddd: hub
hub: fff
eee: dac
dac: fff
fff: ggg hhh
ggg: out
hhh: out
"""
```

```python
from functools import cache

# https://gitlab.com/0xdf/aoc2025/-/blob/main/day11/claude-day11.py?ref_type=heads

# Part 2: Count paths from 'svr' to 'out' that visit both 'dac' and 'fft'
# Use bitmask: 0=none, 1=dac, 2=fft, 3=both
@cache
def count_paths2(node, visited_mask):
    if node == 'dac':
        visited_mask |= 1
    if node == 'fft':
        visited_mask |= 2

    if node == 'out':
        return 1 if visited_mask == 3 else 0
    if node not in graph:
        return 0

    return sum(count_paths2(n, visited_mask) for n in graph[node])

graph = {}

def solution2(data):
    global graph
    lines = data.split("\n")[:-1]    
    
    output = 0
    graph = {}
    for line in lines:
        key, *values = line.split(" ")
        graph[key[:-1]] = values
    count_paths2.cache_clear()
    return count_paths2('svr', 0)


t0 = time.time()
sol2 = solution2(test_data)
print(f"elapsed time {time.time()-t0:.2f} s")
print(sol2)

assert sol2 == test_result2
print("test passed")
```

```python
with open(f"Day{day:02d}.txt") as f:
    inp_data = f.read()

t0 = time.time()
sol2 = solution2(inp_data)
print(f"elapsed time {time.time()-t0:.6f} s")

print(sol2)

assert sol2 == 337433554149492
print("test passed")
```
